## Steps

### 1. Python and NumPy

```python
import numpy as np

prices = [100.0, 101.0, 103.0]
price_array = np.array(prices)
simple_returns = price_array[1:] / price_array[:-1] - 1
print('Python prices:', prices)
print('NumPy simple returns:', np.round(simple_returns, 6))
```

### 2. pandas

```python
import pandas as pd

price_frame = pd.DataFrame({
    'Trade Date': ['2026-01-05', '2026-01-02', '2026-01-06'],
    'Adj Close ($)': [101.0, 100.0, 103.0],
})
price_frame
```

### 3. Reusable Utilities

```python
from src.utils import clean_columns, parse_date_column

cleaned = clean_columns(price_frame)
cleaned = parse_date_column(cleaned, 'trade_date')
cleaned
```
