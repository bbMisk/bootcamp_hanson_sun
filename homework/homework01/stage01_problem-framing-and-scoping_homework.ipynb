## Draft: Scoping Paragraph

Near-term equity-market risk can change faster than a portfolio review cycle. This project will build a reproducible workflow that estimates SPY's annualized realized volatility over the next five trading days using only information available at the forecast date. Success means producing a leakage-safe forecast that improves on a rolling-volatility baseline on a time-ordered holdout set and can be translated into a clear low/normal/high risk band.

The primary stakeholder and user is a portfolio risk lead at a small asset manager. The useful predictive answer is a numeric five-day volatility forecast, a historical risk band, and a short action cue. A high-risk flag prompts review of exposure and hedging; it does not predict return direction, automate a trade, or constitute investment advice.

## Stakeholder Artifact — Memo

### Audience and cadence
Portfolio risk lead; daily refresh and weekly risk review.

### Decision supported
Decide whether current equity exposure deserves closer review or hedging attention.

### What the stakeholder receives
- A five-trading-day annualized realized-volatility forecast.
- A low/normal/high historical risk band.
- A plain-language action cue and assumptions.
- Time-ordered holdout evidence against a rolling-volatility baseline.

### Assumptions, constraints, and risks
- SPY proxies broad U.S. equity exposure and adjusted daily prices are adequate for version one.
- Every feature must be observable at the forecast date.
- Regime changes, abrupt events, and vendor revisions can reduce usefulness.
- The output supports human review and never places a trade.
